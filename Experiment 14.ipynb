```python
%%writefile graphics/rectangle.py
def area(l, b):
    return l * b

def perimeter(l, b):
    return 2 * (l + b)
```

```python
%%writefile graphics/circle.py
import math

def area(r):
    return math.pi * r * r

def perimeter(r):
    return 2 * math.pi * r
```

```python
%%writefile graphics/graphics3D/cuboid.py
def area(l, b, h):
    return 2 * ((l * b) + (l * h) + (b * h))

def perimeter(l, b, h):
    return 4 * (l + b + h)
```

```python
%%writefile graphics/graphics3D/sphere.py
import math

def area(r):
    return 4 * math.pi * r * r

def perimeter(r):
    # Circumference of the sphere's great circle
    return 2 * math.pi * r
```

```python
# Find area, perimeter, surface area and circumference also use different import function

from graphics import circle, rectangle
from graphics.graphics3D import *

while True:
    print("1. Rectangle")
    print("2. Circle")
    print("3. Cuboid")
    print("4. Sphere")
    print("5. Exit")

    ch = int(input("Enter your choice: "))

    if ch == 1:
        l = float(input("Enter length: "))
        b = float(input("Enter breadth: "))

        print("Area:", rectangle.area(l, b))
        print("Perimeter:", rectangle.perimeter(l, b))

    elif ch == 2:
        r = float(input("Enter radius: "))

        print("Area:", round(circle.area(r), 2))
        print("Perimeter:", round(circle.perimeter(r), 2))

    elif ch == 3:
        l = float(input("Enter length: "))
        b = float(input("Enter breadth: "))
        h = float(input("Enter height: "))

        print("Total Surface Area:", cuboid.area(l, b, h))
        print("Perimeter:", cuboid.perimeter(l, b, h))

    elif ch == 4:
        r = float(input("Enter radius: "))

        print("Surface Area:", round(area(r), 2))
        print("Great Circle Circumference:", round(perimeter(r), 2))

    elif ch == 5:
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```
