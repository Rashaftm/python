```python
#all the colors in set1 but not in set2
set1={"Yellow","Red","Purple","Indigo"}
set2={"Green","Maroon","Yellow"}
diff=set1-set2
print("The colors in a but not b is:",diff)
```

```python
#all the colors in set1 and set2 combined
set1={"Yellow","Red","Purple","Indigo"}
set2={"Green","Maroon","Yellow"}
union=set1.union(set2)
print("The colors in set1 and set2 combined is:",union)
```
