```python
strings=input("Enter a string:")
print("The string is:",strings)
if len(strings)>=3:
    if strings.endswith("ing"):
        strings=strings[:-3]+"ly"
        print("the string now is",strings)
    else:
        strings=strings+"ing"
        print("the string now is",strings)
```

```python
strings = input("Enter a string: ")
print("The string is:",strings)
if strings == strings[-1::-1]:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```
