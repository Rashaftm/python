```python
#return the longest word count
def longest_word(words):
    return max(len(word) for word in words)

words = input("Enter words separated by spaces: ").split()

print("Length of the longest word:", longest_word(words))
```

```python
#pattern using nested loop
n = int(input("Enter the number of lines: "))
pattern = ["* " * i for i in range(1, n + 1)] + \
          ["* " * i for i in range(n - 1, 0, -1)]
print("\n".join(pattern))
```

```python
#Factors of a number
n = int(input("Enter a number: "))
factors = [i for i in range(1, n + 1) if n % i == 0]
print("Factors of", n, "are:", factors)
```

```python
#area of square,rectangle and triangle using lambda
square = lambda side: side * side

rectangle = lambda length, breadth: length * breadth

triangle = lambda base, height: 0.5 * base * height

s = float(input("Enter the side of the square: "))
l = float(input("Enter the length of the rectangle: "))
b = float(input("Enter the breadth of the rectangle: "))
base = float(input("Enter the base of the triangle: "))
height = float(input("Enter the height of the triangle: "))

print("Area of square:", square(s))
print("Area of rectangle:", rectangle(l, b))
print("Area of triangle:", triangle(base, height))
```
