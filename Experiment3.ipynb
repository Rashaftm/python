```python
#Reverse a given string
s=input("Enter a string:")
new_string=s[-1]+s[1:-1]+s[0]
print("New string is:",new_string)
```

```python
#String made of first 2 and the last 2 char from a string
s=input("Enter a string:")
new_string=s[0:3]+s[-2::1]
print("New string is:",new_string)
```

```python
#create single string separated with space from two string by swapping the character at position 1
s1=input("Enter the first string:")
s2=input("Enter the second string:")
new_s1=s2[0]+s1[1::1]
new_s2=s1[0]+s2[1::1]
print("New string is:",new_s1," ",new_s2)
```

```python
#Get string from a given string where all occurences of its first char have been changed $, except the first char itself
s=input("Enter the string:")
new_string=s[0]+s[1::1].replace(s[0],"$")
print("New string is:",new_string)
```
