```python
#rectangle class to find area and compare
class Rectangle:
    def get(self, l, b):
        self.l = l
        self.b = b
    def area(self):
        return self.l * self.b
    def perimeter(self):
        return 2 * (self.l + self.b)
r1 = Rectangle()
r2 = Rectangle()
l1 = int(input("Enter length of Rectangle 1: "))
b1 = int(input("Enter breadth of Rectangle 1: "))

l2 = int(input("Enter length of Rectangle 2: "))
b2 = int(input("Enter breadth of Rectangle 2: "))

r1.get(l1, b1)
r2.get(l2, b2)
p = r1.area()
q = r2.area()
print("Area of Rectangle 1:", p)
print("Area of Rectangle 2:", q)

if p > q:
    print("Rectangle 1 is greater")
elif p < q:
    print("Rectangle 2 is greater")
else:
    print("Rectangles are equal")
```

```python
#Using constructor find amount to deposit, withdraw and display
class bankaccount:
    def __init__(self, a_name, a_no, a_type, a_balance):
        self.acc_name = a_name
        self.acc_no = a_no
        self.acc_type = a_type
        self.acc_balance = a_balance
    def deposit(self, a_deposit):
        print("Initial balance:", self.acc_balance)
        self.acc_balance += a_deposit
        print(a_deposit, "is deposited")
        print("Current balance:", self.acc_balance)
    def withdraw(self):
        self.amount = int(input("Withdraw amount: "))
        print("Initial balance:", self.acc_balance)

        if self.amount > self.acc_balance:
            print("You don't have enough balance to withdraw!!")
            print("Current balance:", self.acc_balance)
        else:
            print(self.amount, "is withdrawn")
            self.acc_balance -= self.amount
            print("Current balance:", self.acc_balance)
    def acc_info(self):
        print("Account holder Name:", self.acc_name)
        print("Account Number:", self.acc_no)
        print("Account Type:", self.acc_type)
        print("Account Balance:", self.acc_balance)
name = input("Enter name: ")
no = int(input("Enter account number: "))
type = input("Enter account type: ")
bal = int(input("Enter initial balance: "))
holder = bankaccount(name, no, type, bal)
while True:
    ch = int(input("\n1. Deposit\n2. Withdraw\n3. Account Info\n4. Exit\nChoose your option: "))
    if ch == 1:
        amount = int(input("Deposit amount: "))
        holder.deposit(amount)
    elif ch == 2:
        holder.withdraw()
    elif ch == 3:
        holder.acc_info()
    elif ch == 4:
        print("Exiting...")
        break
    else:
        print("Invalid choice!")
```

```python
#multilevel inheritance using the classes publisher, book, and python.
class publisher:
    def __init__(self, pname):
        self.pubname = pname

    def display1(self):
        print("Book details")
        print("Publisher name:", self.pubname)
class book(publisher):
    def __init__(self, pname, bname, author):
        self.title = bname
        self.auth = author
        publisher.__init__(self, pname)
    def display2(self):
        print("Title:", self.title)
        print("Author:", self.auth)
class python(book):
    def __init__(self, pname, title, au, price, np):
        self.rate = price
        self.nop = np
        book.__init__(self, pname, title, au)
    def display3(self):
        print("Rate:", self.rate)
        print("No. of pages:", self.nop)
p = input("Enter publisher name: ")
b = input("Enter book name: ")
a = input("Enter author name: ")
r = int(input("Enter rate of book: "))
n = int(input("Enter no. of pages: "))
obj = python(p, b, a, r, n)
obj.display1()
obj.display2()
obj.display3()
```

```python
#Compares the areas of two rectangles using operator overloading 
class rectangle:
    def __init__(self, l, w):
        self.length = l
        self.width = w
        self.a = l * w
    def __lt__(self, other):
        if self.a < other.a:
            print("Rectangle 2 is greater")
        elif self.a > other.a:
            print("Rectangle 1 is greater")
        else:
            print("Both rectangles are equal")
l1 = int(input("Enter length of rectangle 1: "))
w1 = int(input("Enter width of rectangle 1: "))

l2 = int(input("Enter length of rectangle 2: "))
w2 = int(input("Enter width of rectangle 2: "))
r1 = rectangle(l1, w1)
r2 = rectangle(l2, w2)
r1 < r2
```

```python
#Operator overloading using the + operator to add two time objects.
class Time:
    def __init__(self, hour, minute, second):
        minute += second // 60
        second = second % 60

        hour += minute // 60
        minute = minute % 60

        self.__hour = hour
        self.__minute = minute
        self.__second = second
    def __add__(self, other):
        h = self.__hour + other.__hour
        m = self.__minute + other.__minute
        s = self.__second + other.__second

        return Time(h, m, s)
    def display(self):
        print("Time:", self.__hour, "hours",
              self.__minute, "minutes",
              self.__second, "seconds")
h1 = int(input("Enter hours of first time: "))
m1 = int(input("Enter minutes of first time: "))
s1 = int(input("Enter seconds of first time: "))
h2 = int(input("Enter hours of second time: "))
m2 = int(input("Enter minutes of second time: "))
s2 = int(input("Enter seconds of second time: "))
t1 = Time(h1, m1, s1)
t2 = Time(h2, m2, s2)
t3 = t1 + t2
print("\nFirst Time:")
t1.display()
print("\nSecond Time:")
t2.display()
print("\nSum of Two Times:")
t3.display()
```
