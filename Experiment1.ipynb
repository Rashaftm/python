```python
#write a python program that accept an integer N and compute the value of n+nn+nnn
n=input ("Enter a number")
n1=int(n)
n2=int(n*2)
n3=int(n*3)
result=n1+n2+n3
print("The result is",n1,"+",n2,"+",n3,"=",result)
```

```python
#input the radius of a circle and compute the area
import math
radius=float(input("Enter the radius:"))
area=math.pi*radius+radius
print(f"Area of a circle is: {area:2f}")
```

```python
#write a prgram to find the biggest of two numbers
a=int(input("Enter the number:a"))
b=int(input("Enter the number:b"))
c=int(input("Enter the number:c"))
biggest=max(a,b,c)
print("The biggest number is:",biggest)
```
