```python
try:
    name=input("Enter the name:")
    s1=int(input("Enter the mark of subject 1:"))
    s2=int(input("Enter the mark of subject 2:"))
    s3=int(input("Enter the mark of subject 3:"))
    
    if s1<0 or s1>100:
        raise ValueError("The mark of the subject cannot be less than 0 or more than 100")
    if s2<0 or s2>100:
        raise ValueError("The mark of the subject cannot be less than 0 or more than 100")
    if s3<0 or s3>100:
        raise ValueError("The mark of the subject cannot be less than 0 or more than 100")
    total=s1+s2+s3
    avg=total/3
except ValueError as e:
    print("Invalid input detected:",e)
else:
    print("The name is:",name)
    print("Total mark is:",total)
    print("Average mark is:",round(avg, 2))
    
    if total>=300:
        print("The student has A grade")
    elif total>=250:
        print("The student has B grade")
    elif total>=200:
        print("The student has C grade")
    elif total>=150:
        print("The student has D grade")
    else:
        print("The student has F grade")
```
