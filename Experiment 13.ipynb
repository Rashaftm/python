```python
%%writefile rectangle.py

def area(l, b):
    return l * b

def perimeter(l, b):
    return 2 * (l + b)
```

```python
%%writefile circle.py
import math

def area(r):
    return round(math.pi * r * r, 2)

def perimeter(r):
    return round(2 * math.pi * r, 2)
```

```python
#Program to find area and perimter using import statement
import rectangle
import circle
while True:
    print("1.Rectangle")
    print("2.Circle")
    print("3.Exit")
    ch=int(input("Enter your choice:"))
    if ch ==1:
        l=int(input("Enter the length of the Rectangle:"))
        b=int(input("Enter the breadth of the Rectangle:"))
        print("Area of Rectangle =",rectangle.area(l,b))
        print("Perimeter of Rectangle =",rectangle.perimeter(l,b))
        print()
    elif ch ==2:
        r=int(input("Enter the radius of the Circle:"))
        print("Area of Circle =",round(circle.area(r),2))
        print("Perimeter of Circle =",round(circle.perimeter(r),2))
        print()
    elif ch ==3:
        print("Exiting program...")
        break
```
