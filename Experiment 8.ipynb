```python
#Accept 5 sudents roll no and name and sort them in ascending and descending order
print("Enter 5 roll number of students:")
roll=input().split(",")
print("Enter 5 names of students:")
name=input().split(",")
student=dict(zip(map(int,roll),name))
ascending=dict(sorted(student.items()))
descending=dict(sorted(student.items(),reverse=True))
print("The dictionary is:",student)
print("ascending order",ascending)
print("descending order",descending)
```

```python
#accept 2 dictionaries and merge them
print("Enter 5 roll number of students:")
roll=input().split(",")
print("Enter 5 names of students:")
name=input().split(",")
dic1=dict(zip(map(int,roll),name))
print("Enter 3 roll number of students:")
roll=input().split(",")
print("Enter 3 names of students:")
name=input().split(",")
dic2=dict(zip(map(int,roll),name))
dic1.update(dic2)
print("The merged dictionary is:",dic1)
```
