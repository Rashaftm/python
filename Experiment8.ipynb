```python
#Accept 5 sudents roll no and name and sort them in ascending and descending order
student={}
for i in range(5):
    roll=(input("Enter the roll no:"))
    name=(input("Enter the name:"))
    student[roll]=name
    
ascending=dict(sorted(student.items()))
descending=dict(sorted(student.items(),reverse=True))
print("the format:",student)
print("ascending order",ascending)
print("descending order",descending)
```

```python
#accept 2 dictionaries and merge them
dic1={}
for i in range(3):
    roll=(input("Enter the roll no for dic1:"))
    name=(input("Enter the name for dic1:"))
    dic1[roll]=name
dic2={}
for i in range(2):
    roll=(input("Enter the roll no for dic2:"))
    name=(input("Enter the name for dic2:"))
    dic2[roll]=name    
dic1.update(dic2)
print("The merged dictionary is:",dic1)
```
